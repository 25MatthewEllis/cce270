## Exercise 2.2 - Adding and removing items

```python
# Run this cell to create a list of foods.

foodList = ['beef', 'steak', 'eggs', 'pork','bacon','ham','ribs','fish']
```

Add 5 more foods to the list, one at a time, then print the whole list.

```python

foodList.append("pizza")
foodList.append("celery")
foodList.append("cokies")
foodList.append("sub")
foodList.append("kfc chicken")

print(foodList)


```

Insert another food to the middle of the list somewhere and print your new list

```python
foodList.insert(3,"burger")
print(foodList)
```

Remove the 6th item in your list and print out your new list and the removed item. Remember that Python counts from 0, so the 6th item is not at index 6.

```python

foodLisremoved=foodList.pop(5)
print(foodList)
print(foodLisremoved)
```

Now remove the **last** item using pop() with no index inside the parentheses. Print both the item that was removed and the list that is left.

```python

last=foodList.pop()
print(foodList)
print(last)
```

Remove beef from your list and print out your new list

```python
foodLisremoved=foodList.remove("beef")
print(foodList)
```

Use sorted() to build a sorted copy of the list. Print the copy, then print the original list. Did sorted() change the original?

```python

listcopy=sorted(foodList)
print(listcopy)
print(foodList)
```

Now sort the list itself using the sort() method and print it. This change is permanent, so the original order is gone.

```python
listcopy.sort()
print(foodList)

```

Reverse the list and print it out

```python

foodList.reverse()
print(foodList)
```

Print the number of items in the list using len().

```python

num_items=len(foodList)
print(num_items)
```

Print how many times 'eggs' appears in the list using count(). Notice that count() answers a different question than len().

```python
num_eggs=foodList.count("eggs")
print(num_eggs)

```

Find the position of 'eggs' in the list using index() and print it.

```python
egg_pos=foodList.index("eggs")
print(egg_pos)

```

## Exercise 2.3 - Summarizing a list of numbers

*Reading: PCC Chapter 4 - Simple Statistics with a List of Numbers. This is next topic's reading.*

```python
# Run this cell to create a list of task durations, in days.

task_durations = [5, 12, 3, 8, 21, 2, 7]
```

Print the total, the longest, and the shortest duration using sum(), max(), and min().

```python
listtotal=sum(task_durations)
listmax=max(task_durations)
listmin=min(task_durations)

print(listtotal)
print(listmax)
print(listmin)

```

Print the average duration. Hint: divide the total by the number of items in the list.

```python
listaverage=listtotal/len(task_durations)
print(listaverage)

```

Print the position of the longest duration in the list. Hint: use index() together with max(). You will use this idea in the homework to report the name of the longest task.

```python

longposition=task_durations.index(max(task_durations))
print(longposition)
```
