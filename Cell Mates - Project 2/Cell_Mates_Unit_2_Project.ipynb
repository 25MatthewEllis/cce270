```python
# @title Let's Play

#remember to upload our rule book "rulebook.csv" and "results.txt"
import csv

#Determine game history using input file
initial_games = 0
initial_player_wins = 0
initial_computer_wins = 0
initial_ties = 0

with open("results.txt","r") as output_file:
  file_content = output_file.read()
  initial_games = file_content.count("Game ")                                   #counts how many games have been played in the text by looking for the word game
  initial_player_wins = file_content.lower().count("player won")                #counts other values looking for "player won, computer won and tie"
  initial_computer_wins = file_content.lower().count("computer won")            #FOUND "FILE_CONTENT".COUNT USING AI!!! - reads the content in the file
  initial_ties = file_content.lower().count("it was a tie")

#Edit file to start adding new session data (Output File)
with open("results.txt","a") as output_file:
  output_file.write("*******NEXT SESSION*******\n\n")                            #choice to indicate new session

#Create game results dictionary
game_results={}                                                                 #Empty dictionary will store all this information we need for our output file

#Class to keep track of the number of user wins, computer wins, and ties
class wins_counter:                                                             #With the class AI was used to help connect the variables from the class to the rest of our code
  def __init__(self,computer_win,player_win,tie):                               #AI was also used within the class for small clarifications and corrections
      self.computer_win = computer_win
      self.player_win = player_win
      self.tie = tie

  def player_wins(self):                                                        #If this method is called it adds one to the player win counter
    self.player_win +=1
    return self.player_win

  def computer_wins(self):                                                      #If this method is called it adds one to the compter win counter
    self.computer_win +=1
    return self.computer_win                                                    #If this method is called it adds one to the tie counter

  def ties(self):
    self.tie +=1
    return self.tie

#initialize win counter
win_count = wins_counter(0,0,0)                                                 #This is the win counter that uses this class

#Defining the user's choice as a function
def user_choice():
  my_choice = input('Choose your move: ').strip().lower()                       #loop will keep going till user chooses valid answer (strip ensures " rock" is valid even in if has a space)
  while my_choice not in ['rock','paper','scissors']:
    print('ERROR: not a valid option.')
    my_choice = input('Choose your move: ').lower()
  return my_choice

#Defining the computer's choice as a function
def computer_choice():
  import random
  cpu_choice = random.choice(['rock', 'paper', 'scissors'])                     #random.choice FOUND FROM THE INTERNET!! - chooses a random option in the list
  print(f'The computer chose: {cpu_choice.title()}')
  return cpu_choice

#The "game" function
def game(my_choice, cpu_choice):
  winner = None                                                                                     #initialize there's no winner at the start
  with open("rulebook.csv", "r") as input_file:                                                     #open the csv file using with so we dont have to close it
    for row in input_file:                                                                          #loop through the rows
      row_list = row.strip().split(',')                                                             #removes white space: makes and seperates list items by the "," comma

      if row_list[0] == "userchoice":                                                               #skips the first row because the first item is "user choice"
         continue

      if row_list[0].lower() == my_choice and row_list[1].lower() == cpu_choice:                    #checking to see if the game matches up with any rows in the rulebook database
        winner = row_list[2]                                                                        #assigns the winner as the value in the winner collumn (2)
        break

  if winner == 'tie':                                                                               #this code just goes through what to print depending on each situation
    win_count.ties()
    print("It's a tie.")
  elif winner == 'computer':                                                                        #early versions of the game code included a lot MORE if statements
    win_count.computer_wins()                                                                       #e.g. "if i chose rock and cpu chose rock return "tie"" etc.
    print("You lost.")                                                                              #this is quicker because it already has all the outcomes in easy table (csv)
  elif winner == 'player':                                                                          #so it works just by looking up values (like VLOOKUP)
    win_count.player_wins()                                                                         #left with just these 4 situations
    print("You win!")
  else:
    print("This game outcome isn't possible according to the rulebook")                             #technically not needed!

#Update which game we're on
  game_number = initial_player_wins + initial_computer_wins + initial_ties + win_count.player_win + win_count.computer_win + win_count.tie                                   #This simple equation uses the outcomes to update the game_number variable to show which game we are on

#Update the output file
  with open("results.txt","a", newline="") as output_file:                                          #opens the file but doesnt erase (appends), if its a tie it lists the game no., the choices and "its a tie"
    if winner == 'tie':                                                                             #if not a tie itll say "player won" or "computer won"
      output_file.write(f"Game {game_number}:\n Player Choice: {my_choice.capitalize()}\n Computer Choice: {cpu_choice.capitalize()} \n It was a {winner.capitalize()}\n\n")
    else:
      output_file.write(f"Game {game_number}:\n Player Choice: {my_choice.capitalize()}\n Computer Choice: {cpu_choice.capitalize()}\n {winner.capitalize()} won\n\n")


#Playing the Game function
def play():                                                                                          #we called this "run_time" in our pseudocode proposal!
  end_game = False
  while end_game == False:
    user = user_choice()
    cpu = computer_choice()
    game(user, cpu)                                                                                  #runs the game function with our choices
    print()

    #asks to play again
    yes_no = input("Do you want to play again? ").lower()
    while yes_no not in ['yes','no']:
      print('ERROR: Invalid input.')
      yes_no = input("Do you want to play again? ").lower()
    if yes_no == 'yes':
      end_game = False
      print()
    elif yes_no == 'no':
      end_game = True

#run the game
play()


total_player_wins = initial_player_wins + win_count.player_win                                    #adds up wins from input file and current session
total_computer_wins = initial_computer_wins + win_count.computer_win
total_ties = initial_ties + win_count.tie

#Print this session results with correct grammar
player_plural = "" if win_count.player_win == 1 else "s"                                          #checks if any of them are more than one and reflects that in the language (correct grammar ppl)
computer_plural = "" if win_count.computer_win == 1 else "s"
tie_verb = "is" if win_count.tie == 1 else "are"                                                  #need to change a whole word  "is 1 tie" vs "are 2 ties"
tie_plural = "" if win_count.tie == 1 else "s"

print(f'\n***This Session***\nThe player has {win_count.player_win} win{player_plural}. The computer has {win_count.computer_win} win{computer_plural}. There {tie_verb} {win_count.tie} tie{tie_plural}.')

#Print All time records with correct grammar
total_player_plural = "" if total_player_wins == 1 else "s"                                       #I know its a lot of extra code but we need the right grammar for both sessions.
total_computer_plural = "" if total_computer_wins == 1 else "s"
total_tie_verb = "is" if total_ties == 1 else "are"
total_tie_plural = "" if total_ties == 1 else "s"

print(f"\n ***All Time***\n The player has {total_player_wins} win{total_player_plural}. The computer has {total_computer_wins} win{total_computer_plural}. There {total_tie_verb} {total_ties} tie{total_tie_plural}.")
```
